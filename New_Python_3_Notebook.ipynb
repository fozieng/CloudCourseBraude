### Student Details and Favorite Program Form

```python
#@title Select Student & Update Favorite Program { run: "auto" }

student_db = {
    "Fozi": {"age": 25, "major": "SW Engineering", "fav_program": "Python"},
    "Ameer": {"age": 20, "major": "SW Engineering", "fav_program": "C++"}
}

# Form variables
student_name = "Ameer" #@param ["Fozi", "Ameer"]
favorite_program = "Python" #@param {type:"string"}

# Save the favorite program to the student
if student_name == "Fozi":
    student_db["Fozi"]["fav_program"] = favorite_program
elif student_name == "Ameer":
    student_db["Ameer"]["fav_program"] = favorite_program

# Show the selected student's details
print("--- Student Information ---")
print("Name:", student_name)

student_info = student_db[student_name]
print("Age:", student_info["age"])
print("Major:", student_info["major"])
print("Favorite Program:", student_info["fav_program"])

print("\n--- All Students ---")
print(student_db)
```
